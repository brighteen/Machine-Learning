```python
import numpy as np
import random
import matplotlib.pyplot as plt
from tqdm.notebook import tqdm

def create_map_np():
    map_data = np.zeros((10, 10), dtype=int)
    obstacles = [(0, 2), (0, 3),(1, 2), (1, 3), (2, 2), (2, 3), (3, 2), (3, 3),
                 (1, 7), (1, 8), (1, 9),
                 (4, 5), (4, 6), (5, 5), (5, 6), (6, 5), (6, 6),
                 (7, 1), (7, 2), (8, 1), (8, 2),
                 (7, 8), (8, 8),
                 (9, 4), (9, 5)]
    for i, j in obstacles:
        map_data[i, j] = 1
    return map_data

def visualize_map(map_data, path=None, title="Map"):
    plt.figure(figsize=(8, 8))
    plt.imshow(map_data, cmap='binary')
    for i in range(map_data.shape[0] + 1):
        plt.axhline(i - 0.5, color='gray', linestyle='-', alpha=0.3)
        plt.axvline(i - 0.5, color='gray', linestyle='-', alpha=0.3)
    for i in range(map_data.shape[0]):
        for j in range(map_data.shape[1]):
            node_num = i * map_data.shape[1] + j + 1
            plt.text(j, i, f'{node_num}', ha='center', va='center', 
                     color='white' if map_data[i, j] == 1 else 'black')
    if path:
        path_coords = []
        for node in path:
            row = (node - 1) // map_data.shape[1]
            col = (node - 1) % map_data.shape[1]
            path_coords.append((col, row))
        x_coords = [x for x, y in path_coords]
        y_coords = [y for x, y in path_coords]
        plt.plot(x_coords, y_coords, 'r-', linewidth=2)
        plt.plot(x_coords, y_coords, 'bo', markersize=8)
        plt.plot(x_coords[0], y_coords[0], 'go', markersize=10)
        plt.plot(x_coords[-1], y_coords[-1], 'mo', markersize=10)
    plt.title(title)
    plt.grid(True)
    plt.xticks(range(map_data.shape[1]))
    plt.yticks(range(map_data.shape[0]))
    plt.show()

def hash_path(path):
    return hash(tuple(path))

def bresenham_line(x0, y0, x1, y1):
    points = []
    dx = abs(x1 - x0)
    dy = abs(y1 - y0)
    x, y = x0, y0
    sx = -1 if x0 > x1 else 1
    sy = -1 if y0 > y1 else 1
    if dx > dy:
        err = dx / 2.0
        while x != x1:
            points.append((x, y))
            err -= dy
            if err < 0:
                y += sy
                err += dx
            x += sx
        points.append((x, y))
    else:
        err = dy / 2.0
        while y != y1:
            points.append((x, y))
            err -= dx
            if err < 0:
                x += sx
                err += dy
            y += sy
        points.append((x, y))
    return points
```

```python
def random_walk_np(start, goal, map_data, max_steps=60):
    current = start
    path = [current]
    steps = 0
    visited = set([current])
    while current != goal and steps < max_steps:
        row, col = (current - 1) // 10, (current - 1) % 10
        possible_moves = []
        for dr in [-1, 0, 1]:
            for dc in [-1, 0, 1]:
                if dr == 0 and dc == 0:
                    continue
                new_row, new_col = row + dr, col + dc
                if (0 <= new_row < 10 and 0 <= new_col < 10 and map_data[new_row, new_col] == 0):
                    new_node = new_row * 10 + new_col + 1
                    if new_node not in visited:
                        possible_moves.append(new_node)
        if not possible_moves:
            if len(path) <= 1:
                return None
            path.pop()
            current = path[-1]
            steps += 1
            continue
        if goal in possible_moves:
            path.append(goal)
            break
        next_node = random.choice(possible_moves)
        path.append(next_node)
        visited.add(next_node)
        current = next_node
        steps += 1
    if current != goal:
        return None
    return path

def bfs_path(start, goal, map_data):
    from collections import deque
    queue = deque()
    queue.append((start, [start]))
    visited = set([start])
    while queue:
        current, p = queue.popleft()
        if current == goal:
            return p
        row, col = (current - 1) // 10, (current - 1) % 10
        for dr in [-1, 0, 1]:
            for dc in [-1, 0, 1]:
                if dr == 0 and dc == 0:
                    continue
                new_row, new_col = row + dr, col + dc
                if 0 <= new_row < 10 and 0 <= new_col < 10 and map_data[new_row, new_col] == 0:
                    next_node = new_row * 10 + new_col + 1
                    if next_node not in visited:
                        visited.add(next_node)
                        queue.append((next_node, p + [next_node]))
    return None

def initialize_population_np(start, goal, map_data, pop_size=10, fixed_length=None):
    population = []
    hash_set = set()
    tries = 0
    while len(population) < pop_size and tries < pop_size * 200:
        path = random_walk_np(start, goal, map_data)
        if path and (fixed_length is None or len(path) == fixed_length):
            h = hash_path(path)
            if h not in hash_set:
                population.append(path)
                hash_set.add(h)
        tries += 1
    # 부족하면 BFS로 채움
    while len(population) < pop_size:
        path = bfs_path(start, goal, map_data)
        if path and (fixed_length is None or len(path) == fixed_length):
            h = hash_path(path)
            if h not in hash_set:
                population.append(path)
                hash_set.add(h)
        else:
            # fallback
            population.append([start, goal])
    return population

repair_cache = {}
def repair_path(path, map_data):
    path_tuple = tuple(path)
    if path_tuple in repair_cache:
        return repair_cache[path_tuple]

    def bfs_path_local(start, goal, map_data):
        from collections import deque
        queue = deque([(start, [start])])
        visited = {start}
        while queue:
            current, p = queue.popleft()
            if current == goal:
                return p
            row, col = (current - 1) // 10, (current - 1) % 10
            # 8가지 방향 (대각선 포함)
            for dr in [-1, 0, 1]:
                for dc in [-1, 0, 1]:
                    if dr == 0 and dc == 0: continue
                    new_row, new_col = row + dr, col + dc
                    if 0 <= new_row < 10 and 0 <= new_col < 10 and map_data[new_row, new_col] == 0:
                        next_node = new_row * 10 + new_col + 1
                        if next_node not in visited:
                            visited.add(next_node)
                            queue.append((next_node, p + [next_node]))
        return None # 경로를 찾지 못한 경우

    # 경로 압축 및 중복 제거
    unique_path = []
    seen = set()
    for node in path:
        if node not in seen:
            unique_path.append(node)
            seen.add(node)
    path = unique_path

    fixed = [path[0]]
    i = 0
    while i < len(path) - 1:
        j = len(path) - 1
        best_next_i = i + 1
        while j > i:
            r0, c0 = (path[i] - 1) // 10, (path[i] - 1) % 10
            r1, c1 = (path[j] - 1) // 10, (path[j] - 1) % 10
            line_points = bresenham_line(r0, c0, r1, c1)
            is_blocked = any(map_data[r, c] == 1 for r, c in line_points)
            if not is_blocked:
                fixed.append(path[j])
                best_next_i = j
                break
            j -= 1
        
        if best_next_i == i + 1: # 직선 경로를 찾지 못한 경우
            sub_path = bfs_path_local(path[i], path[i+1], map_data)
            if sub_path and len(sub_path) > 1:
                for node in sub_path[1:]:
                     if node not in fixed:
                        fixed.append(node)
            else: # BFS도 실패하면 그냥 다음 노드로 이동
                 if path[i+1] not in fixed:
                    fixed.append(path[i+1])
        i = best_next_i

    repair_cache[path_tuple] = fixed
    return fixed

class PathEvaluator:
    def __init__(self, map_data):
        self.map_data = map_data
        self.cache = {}

    def evaluate(self, path):
        path_tuple = tuple(path)
        if path_tuple in self.cache:
            return self.cache[path_tuple]

        # 경로 유효성 검사
        for i in range(len(path) - 1):
            r0, c0 = (path[i] - 1) // 10, (path[i] - 1) % 10
            r1, c1 = (path[i+1] - 1) // 10, (path[i+1] - 1) % 10
            line_points = bresenham_line(r0, c0, r1, c1)
            if any(self.map_data[r, c] == 1 for r, c in line_points):
                self.cache[path_tuple] = float('inf')
                return float('inf')

        # 거리 계산 (벡터화)
        if len(path) < 2:
            self.cache[path_tuple] = 0.0
            return 0.0
            
        coords = np.array([( (n-1)//10, (n-1)%10 ) for n in path])
        diffs = np.diff(coords, axis=0)
        distances = np.sqrt(np.sum(diffs**2, axis=1))
        total_dist = np.sum(distances)
        
        self.cache[path_tuple] = total_dist
        return total_dist

    def batch_evaluate(self, population):
        return [self.evaluate(p) for p in population]
```

```python
class ProposedGA_Repair:
    def __init__(self, map_data, start, goal, pop_size=10, max_gen=15):
        self.map_data = map_data
        self.start = start
        self.goal = goal
        self.pop_size = pop_size
        self.max_gen = max_gen
        self.crossover_rate = 0.7
        self.mutation_rate = 0.2
        self.shortcut_rate = 0.3
        self.population = initialize_population_np(start, goal, map_data, pop_size)
        self.evaluator = PathEvaluator(map_data)
        self.best_path = None
        self.best_distance = float('inf')
        self.best_history = []

    def crossover(self, path1, path2):
        minlen = min(len(path1), len(path2))
        if minlen <= 2:
            return path1.copy()
        point = random.randint(1, minlen - 2)
        new_path = path1[:point] + path2[point:]
        dedup_path = []
        seen = set()
        for n in new_path:
            if n not in seen:
                dedup_path.append(n)
                seen.add(n)
        if dedup_path[-1] != self.goal:
            dedup_path.append(self.goal)
        repaired = repair_path(dedup_path, self.map_data)
        return repaired if repaired is not None else [self.start, self.goal]

    def mutation(self, path):
        if len(path) <= 2:
            return path.copy()
        point = random.randint(1, len(path) - 2)
        row, col = (path[point] - 1) // 10, (path[point] - 1) % 10
        possible_nodes = []
        for dr in [-1, 0, 1]:
            for dc in [-1, 0, 1]:
                new_row, new_col = row + dr, col + dc
                new_node = new_row * 10 + new_col + 1
                if (0 <= new_row < 10 and 0 <= new_col < 10 and 
                    self.map_data[new_row, new_col] == 0 and new_node not in path):
                    possible_nodes.append(new_node)
        if not possible_nodes:
            return path.copy()
        new_path = path.copy()
        new_path[point] = random.choice(possible_nodes)
        dedup_path = []
        seen = set()
        for n in new_path:
            if n not in seen:
                dedup_path.append(n)
                seen.add(n)
        if dedup_path[-1] != self.goal:
            dedup_path.append(self.goal)
        repaired = repair_path(dedup_path, self.map_data)
        return repaired if repaired is not None else [self.start, self.goal]

    def shortcut(self, path):
        if len(path) <= 3:
            return path.copy()
        l = len(path)
        idx1, idx2 = sorted(random.sample(range(l), 2))
        if idx2 - idx1 > 1:
            new_path = path[:idx1+1] + path[idx2:]
            dedup_path = []
            seen = set()
            for n in new_path:
                if n not in seen:
                    dedup_path.append(n)
                    seen.add(n)
            if dedup_path[-1] != self.goal:
                dedup_path.append(self.goal)
            repaired = repair_path(dedup_path, self.map_data)
            return repaired if repaired is not None else [self.start, self.goal]
        return path.copy()

    def evolve(self):
        for gen in tqdm(range(self.max_gen), desc="ProposedGA 진행"):
            uniq_hash = set()
            uniq_pop = []
            for path in self.population:
                h = hash_path(path)
                if h not in uniq_hash:
                    uniq_pop.append(path)
                    uniq_hash.add(h)
            if not uniq_pop:
                uniq_pop.append([self.start, self.goal])
            self.population = uniq_pop

            new_population = []
            tries = 0
            while len(new_population) < self.pop_size and tries < self.pop_size * 5:
                parent = random.choice(self.population)
                child = parent.copy()
                if random.random() < self.crossover_rate:
                    parent2 = random.choice(self.population)
                    child = self.crossover(child, parent2)
                if random.random() < self.mutation_rate:
                    child = self.mutation(child)
                if random.random() < self.shortcut_rate:
                    child = self.shortcut(child)
                h = hash_path(child)
                if h not in uniq_hash and self.evaluator.evaluate(child) != float('inf'):
                    new_population.append(child)
                    uniq_hash.add(h)
                tries += 1

            if not new_population:
                new_population.append([self.start, self.goal])

            all_population = self.population + new_population
            valid_population = [p for p in all_population if self.evaluator.evaluate(p) != float('inf')]
            if not valid_population:
                valid_population = [[self.start, self.goal]]
            scores = self.evaluator.batch_evaluate(valid_population)
            scores_np = np.array(scores)
            sorted_indices = np.argsort(scores_np)
            self.population = [valid_population[i] for i in sorted_indices[:self.pop_size]]
            current_best = scores_np[sorted_indices[0]]
            current_mean = scores_np[:self.pop_size].mean()
            if current_best < self.best_distance:
                self.best_path = self.population[0]
                self.best_distance = current_best
            self.best_history.append(self.best_distance)
            if gen % 3 == 0 or gen == self.max_gen - 1:
                print(f"[세대 {gen+1:02d}] 최단 거리: {current_best:.2f} / 평균 거리: {current_mean:.2f} / 고유경로수: {len(self.population)}")
        return self.best_path, self.best_distance
```

```python
class FGA:
    def __init__(self, map_data, start, goal, pop_size=10, max_gen=15, length=15):
        self.map_data = map_data
        self.start = start
        self.goal = goal
        self.length = length
        self.pop_size = pop_size
        self.max_gen = max_gen
        self.crossover_rate = 0.7
        self.mutation_rate = 0.2
        self.shortcut_rate = 0.3
        self.population = initialize_population_np(start, goal, map_data, pop_size, fixed_length=length)
        self.evaluator = PathEvaluator(map_data)
        self.best_path = None
        self.best_distance = float('inf')
        self.best_history = []

    def crossover(self, path1, path2):
        point = random.randint(1, self.length - 2)
        new_path = path1[:point] + path2[point:]
        if new_path[-1] != self.goal:
            new_path[-1] = self.goal
        dedup = []
        seen = set()
        for n in new_path:
            if n not in seen:
                dedup.append(n)
                seen.add(n)
        while len(dedup) < self.length:
            dedup.insert(-1, random.randint(1, 100))
        if len(dedup) > self.length:
            dedup = dedup[:self.length - 1] + [self.goal]
        repaired = repair_path(dedup, self.map_data)
        return repaired if repaired is not None else [self.start, self.goal]

    def mutation(self, path):
        new_path = path.copy()
        point = random.randint(1, self.length - 2)
        row, col = (new_path[point] - 1) // 10, (new_path[point] - 1) % 10
        possible_nodes = []
        for dr in [-1, 0, 1]:
            for dc in [-1, 0, 1]:
                new_row, new_col = row + dr, col + dc
                new_node = new_row * 10 + new_col + 1
                if (0 <= new_row < 10 and 0 <= new_col < 10 and
                    self.map_data[new_row, new_col] == 0 and new_node not in new_path):
                    possible_nodes.append(new_node)
        if possible_nodes:
            new_path[point] = random.choice(possible_nodes)
        repaired = repair_path(new_path, self.map_data)
        return repaired if repaired is not None else [self.start, self.goal]

    def shortcut(self, path):
        # FGA에서는 길이 유지, 중복제거만 적용
        return path.copy() # 생략

    def evolve(self):
        for gen in tqdm(range(self.max_gen), desc="FGA 진행"):
            uniq_hash = set()
            uniq_pop = []
            for path in self.population:
                h = hash_path(path)
                if h not in uniq_hash:
                    uniq_pop.append(path)
                    uniq_hash.add(h)
            if not uniq_pop:
                uniq_pop.append([self.start, self.goal])
            self.population = uniq_pop

            new_population = []
            tries = 0
            while len(new_population) < self.pop_size and tries < self.pop_size * 5:
                parent = random.choice(self.population)
                child = parent.copy()
                if random.random() < self.crossover_rate:
                    parent2 = random.choice(self.population)
                    child = self.crossover(child, parent2)
                if random.random() < self.mutation_rate:
                    child = self.mutation(child)
                h = hash_path(child)
                if h not in uniq_hash and self.evaluator.evaluate(child) != float('inf'):
                    new_population.append(child)
                    uniq_hash.add(h)
                tries += 1

            if not new_population:
                new_population.append([self.start, self.goal])

            all_population = self.population + new_population
            valid_population = [p for p in all_population if self.evaluator.evaluate(p) != float('inf')]
            if not valid_population:
                valid_population = [[self.start, self.goal]]
            scores = self.evaluator.batch_evaluate(valid_population)
            scores_np = np.array(scores)
            sorted_indices = np.argsort(scores_np)
            self.population = [valid_population[i] for i in sorted_indices[:self.pop_size]]
            current_best = scores_np[sorted_indices[0]]
            current_mean = scores_np[:self.pop_size].mean()
            if current_best < self.best_distance:
                self.best_path = self.population[0]
                self.best_distance = current_best
            self.best_history.append(self.best_distance)
            if gen % 3 == 0 or gen == self.max_gen - 1:
                print(f"[세대 {gen+1:02d}] 최단 거리: {current_best:.2f} / 평균 거리: {current_mean:.2f} / 고유경로수: {len(self.population)}")
        return self.best_path, self.best_distance
```

```python
class IGA:
    def __init__(self, map_data, start, goal, pop_size=10, max_gen=15):
        self.map_data = map_data
        self.start = start
        self.goal = goal
        self.pop_size = pop_size
        self.max_gen = max_gen
        self.crossover_rate = 0.7
        self.mutation_rate = 0.2
        self.shortcut_rate = 0.3
        self.population = initialize_population_np(start, goal, map_data, pop_size)
        self.evaluator = PathEvaluator(map_data)
        self.best_path = None
        self.best_distance = float('inf')
        self.best_history = []

    def crossover(self, path1, path2):
        # 임의의 지점에서 삽입/삭제도 허용
        cut1 = random.randint(1, len(path1)-2)
        cut2 = random.randint(1, len(path2)-2)
        new_path = path1[:cut1] + path2[cut2:]
        dedup_path = []
        seen = set()
        for n in new_path:
            if n not in seen:
                dedup_path.append(n)
                seen.add(n)
        if dedup_path[-1] != self.goal:
            dedup_path.append(self.goal)
        repaired = repair_path(dedup_path, self.map_data)
        return repaired if repaired is not None else [self.start, self.goal]

    def mutation(self, path):
        new_path = path.copy()
        # 삽입, 삭제, 변경 중 하나 랜덤 선택
        op = random.choice(['insert', 'delete', 'change'])
        if op == 'insert' and len(new_path) < 20:
            # 임의 위치에 인접노드 삽입
            idx = random.randint(1, len(new_path)-1)
            node = new_path[idx-1]
            row, col = (node - 1) // 10, (node - 1) % 10
            candidates = []
            for dr in [-1, 0, 1]:
                for dc in [-1, 0, 1]:
                    new_row, new_col = row + dr, col + dc
                    new_node = new_row * 10 + new_col + 1
                    if (0 <= new_row < 10 and 0 <= new_col < 10 and 
                        self.map_data[new_row, new_col] == 0 and new_node not in new_path):
                        candidates.append(new_node)
            if candidates:
                new_path.insert(idx, random.choice(candidates))
        elif op == 'delete' and len(new_path) > 3:
            idx = random.randint(1, len(new_path)-2)
            del new_path[idx]
        else: # change
            idx = random.randint(1, len(new_path)-2)
            node = new_path[idx]
            row, col = (node - 1) // 10, (node - 1) % 10
            candidates = []
            for dr in [-1, 0, 1]:
                for dc in [-1, 0, 1]:
                    new_row, new_col = row + dr, col + dc
                    new_node = new_row * 10 + new_col + 1
                    if (0 <= new_row < 10 and 0 <= new_col < 10 and 
                        self.map_data[new_row, new_col] == 0 and new_node not in new_path):
                        candidates.append(new_node)
            if candidates:
                new_path[idx] = random.choice(candidates)
        if new_path[-1] != self.goal:
            new_path[-1] = self.goal
        repaired = repair_path(new_path, self.map_data)
        return repaired if repaired is not None else [self.start, self.goal]

    def shortcut(self, path):
        if len(path) <= 3:
            return path.copy()
        l = len(path)
        idx1, idx2 = sorted(random.sample(range(l), 2))
        if idx2 - idx1 > 1:
            new_path = path[:idx1+1] + path[idx2:]
            dedup_path = []
            seen = set()
            for n in new_path:
                if n not in seen:
                    dedup_path.append(n)
                    seen.add(n)
            if dedup_path[-1] != self.goal:
                dedup_path.append(self.goal)
            repaired = repair_path(dedup_path, self.map_data)
            return repaired if repaired is not None else [self.start, self.goal]
        return path.copy()

    def evolve(self):
        for gen in tqdm(range(self.max_gen), desc="IGA 진행"):
            uniq_hash = set()
            uniq_pop = []
            for path in self.population:
                h = hash_path(path)
                if h not in uniq_hash:
                    uniq_pop.append(path)
                    uniq_hash.add(h)
            if not uniq_pop:
                uniq_pop.append([self.start, self.goal])
            self.population = uniq_pop

            new_population = []
            tries = 0
            while len(new_population) < self.pop_size and tries < self.pop_size * 5:
                parent = random.choice(self.population)
                child = parent.copy()
                if random.random() < self.crossover_rate:
                    parent2 = random.choice(self.population)
                    child = self.crossover(child, parent2)
                if random.random() < self.mutation_rate:
                    child = self.mutation(child)
                if random.random() < self.shortcut_rate:
                    child = self.shortcut(child)
                h = hash_path(child)
                if h not in uniq_hash and self.evaluator.evaluate(child) != float('inf'):
                    new_population.append(child)
                    uniq_hash.add(h)
                tries += 1

            if not new_population:
                new_population.append([self.start, self.goal])

            all_population = self.population + new_population
            valid_population = [p for p in all_population if self.evaluator.evaluate(p) != float('inf')]
            if not valid_population:
                valid_population = [[self.start, self.goal]]
            scores = self.evaluator.batch_evaluate(valid_population)
            scores_np = np.array(scores)
            sorted_indices = np.argsort(scores_np)
            self.population = [valid_population[i] for i in sorted_indices[:self.pop_size]]
            current_best = scores_np[sorted_indices[0]]
            current_mean = scores_np[:self.pop_size].mean()
            if current_best < self.best_distance:
                self.best_path = self.population[0]
                self.best_distance = current_best
            self.best_history.append(self.best_distance)
            if gen % 3 == 0 or gen == self.max_gen - 1:
                print(f"[세대 {gen+1:02d}] 최단 거리: {current_best:.2f} / 평균 거리: {current_mean:.2f} / 고유경로수: {len(self.population)}")
        return self.best_path, self.best_distance
```

```python
class KGA:
    def __init__(self, map_data, start, goal, pop_size=10, max_gen=15, keypoints=3):
        self.map_data = map_data
        self.start = start
        self.goal = goal
        self.keypoints = keypoints
        self.pop_size = pop_size
        self.max_gen = max_gen
        self.crossover_rate = 0.7
        self.mutation_rate = 0.2
        self.shortcut_rate = 0.3
        self.population = self.init_population()
        self.evaluator = PathEvaluator(map_data)
        self.best_path = None
        self.best_distance = float('inf')
        self.best_history = []

    def init_population(self):
        # keypoints 개수만큼 랜덤 중간 노드 → start~key1~key2...~goal 경로를 BFS로 연결
        population = []
        tries = 0
        while len(population) < self.pop_size and tries < self.pop_size * 50:
            keys = random.sample(range(2, 99), self.keypoints)
            via = [self.start] + sorted(keys) + [self.goal]
            path = []
            fail = False
            for i in range(len(via)-1):
                seg = bfs_path(via[i], via[i+1], self.map_data)
                if seg is None:
                    fail = True
                    break
                if i > 0:
                    seg = seg[1:] # 중복 연결노드 삭제
                path += seg
            if not fail:
                h = hash_path(path)
                if h not in [hash_path(p) for p in population]:
                    population.append(path)
            tries += 1
        while len(population) < self.pop_size:
            population.append([self.start, self.goal])
        return population

    def crossover(self, path1, path2):
        # keypoints만 교차(경유지점)
        kp1 = [path1[0]] + sorted(random.sample(path1[1:-1], min(self.keypoints, len(path1)-2))) + [path1[-1]]
        kp2 = [path2[0]] + sorted(random.sample(path2[1:-1], min(self.keypoints, len(path2)-2))) + [path2[-1]]
        via = []
        for k1, k2 in zip(kp1[1:-1], kp2[1:-1]):
            via.append(random.choice([k1, k2]))
        via = [self.start] + via + [self.goal]
        path = []
        fail = False
        for i in range(len(via)-1):
            seg = bfs_path(via[i], via[i+1], self.map_data)
            if seg is None:
                fail = True
                break
            if i > 0:
                seg = seg[1:]
            path += seg
        if fail or not path:
            return [self.start, self.goal]
        return path

    def mutation(self, path):
        # keypoints 하나 랜덤하게 바꿈
        if len(path) <= self.keypoints+2:
            return path.copy()
        kp = [path[0]] + sorted(random.sample(path[1:-1], min(self.keypoints, len(path)-2))) + [path[-1]]
        idx = random.randint(1, self.keypoints)
        new_kp = random.randint(2, 99)
        kp[idx] = new_kp
        via = kp
        path2 = []
        fail = False
        for i in range(len(via)-1):
            seg = bfs_path(via[i], via[i+1], self.map_data)
            if seg is None:
                fail = True
                break
            if i > 0:
                seg = seg[1:]
            path2 += seg
        if fail or not path2:
            return [self.start, self.goal]
        return path2

    def shortcut(self, path):
        if len(path) <= 3:
            return path.copy()
        l = len(path)
        idx1, idx2 = sorted(random.sample(range(l), 2))
        if idx2 - idx1 > 1:
            new_path = path[:idx1+1] + path[idx2:]
            dedup_path = []
            seen = set()
            for n in new_path:
                if n not in seen:
                    dedup_path.append(n)
                    seen.add(n)
            if dedup_path[-1] != self.goal:
                dedup_path.append(self.goal)
            repaired = repair_path(dedup_path, self.map_data)
            return repaired if repaired is not None else [self.start, self.goal]
        return path.copy()

    def evolve(self):
        for gen in tqdm(range(self.max_gen), desc="KGA 진행"):
            uniq_hash = set()
            uniq_pop = []
            for path in self.population:
                h = hash_path(path)
                if h not in uniq_hash:
                    uniq_pop.append(path)
                    uniq_hash.add(h)
            if not uniq_pop:
                uniq_pop.append([self.start, self.goal])
            self.population = uniq_pop

            new_population = []
            tries = 0
            while len(new_population) < self.pop_size and tries < self.pop_size * 5:
                parent = random.choice(self.population)
                child = parent.copy()
                if random.random() < self.crossover_rate:
                    parent2 = random.choice(self.population)
                    child = self.crossover(child, parent2)
                if random.random() < self.mutation_rate:
                    child = self.mutation(child)
                if random.random() < self.shortcut_rate:
                    child = self.shortcut(child)
                h = hash_path(child)
                if h not in uniq_hash and self.evaluator.evaluate(child) != float('inf'):
                    new_population.append(child)
                    uniq_hash.add(h)
                tries += 1

            if not new_population:
                new_population.append([self.start, self.goal])

            all_population = self.population + new_population
            valid_population = [p for p in all_population if self.evaluator.evaluate(p) != float('inf')]
            if not valid_population:
                valid_population = [[self.start, self.goal]]
            scores = self.evaluator.batch_evaluate(valid_population)
            scores_np = np.array(scores)
            sorted_indices = np.argsort(scores_np)
            self.population = [valid_population[i] for i in sorted_indices[:self.pop_size]]
            current_best = scores_np[sorted_indices[0]]
            current_mean = scores_np[:self.pop_size].mean()
            if current_best < self.best_distance:
                self.best_path = self.population[0]
                self.best_distance = current_best
            self.best_history.append(self.best_distance)
            if gen % 3 == 0 or gen == self.max_gen - 1:
                print(f"[세대 {gen+1:02d}] 최단 거리: {current_best:.2f} / 평균 거리: {current_mean:.2f} / 고유경로수: {len(self.population)}")
        return self.best_path, self.best_distance
```
