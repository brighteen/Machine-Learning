```python
class DisjointSet:
    def __init__(self, n):
        self.parent = [i for i in range(n)]
        self.rank = [0] * n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, x, y):
        root_x = self.find(x)
        root_y = self.find(y)
        if root_x != root_y:
            if self.rank[root_x] < self.rank[root_y]:
                self.parent[root_x] = root_y
            elif self.rank[root_x] > self.rank[root_y]:
                self.parent[root_y] = root_x
            else:
                self.parent[root_y] = root_x
                self.rank[root_x] += 1

def kruskal_mst(graph):
    edges = []
    nodes = list(graph.keys())  # 딕셔너리 키들을 리스트로 변경
    num_nodes = len(nodes)
    for i in range(num_nodes):
        for j in range(i + 1, num_nodes):
            src = nodes[i]
            dest = nodes[j]
            if dest in graph[src]:
                cost = graph[src][dest]
                edges.append((cost, src, dest))
    edges.sort()

    disjoint_set = DisjointSet(num_nodes)
    mst = []

    for edge in edges:
        cost, src, dest = edge
        src_index = nodes.index(src)
        dest_index = nodes.index(dest)
        if disjoint_set.find(src_index) != disjoint_set.find(dest_index):
            disjoint_set.union(src_index, dest_index)
            mst.append(edge)

    return mst

graph = {
    'S': {'A': 4, 'B': 3, 'C': 5},
    'A': {'C': 3, 'D': 5},
    'B': {'C': 1, 'E': 4},
    'C': {'E': 3, 'D': 3},
    'D': {'E': 2, 'T': 5},
    'E': {'T': 6},
    'T': {}
}

mst = kruskal_mst(graph)
total_cost = sum(cost for cost, _, _ in mst)
print("최소 비용 신장 트리의 간선 및 비용:")
for edge in mst:
    print(edge)
print("총 설치 비용:", total_cost)
```

```python
import heapq

def dijkstra(graph, start):
    distances = {node: float('inf') for node in graph}
    distances[start] = 0
    queue = [(0, start)]

    while queue:
        current_distance, current_node = heapq.heappop(queue)
        if current_distance > distances[current_node]:
            continue
        for neighbor, weight in graph[current_node].items():
            distance = current_distance + weight
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                heapq.heappush(queue, (distance, neighbor))

    return distances

graph = {
    'S': {'A': 4, 'B': 3, 'C': 5},
    'A': {'S': 4, 'C': 3, 'D': 5},
    'B': {'S': 3, 'C': 1, 'E': 4},
    'C': {'S': 5, 'A': 3, 'B': 1, 'D': 3, 'E': 3},
    'D': {'A': 5, 'C': 3, 'E': 2, 'T': 5},
    'E': {'B': 4, 'C': 3, 'D': 2, 'T': 6},
    'T': {'D': 5, 'E': 6}
}

start_node = 'S'
shortest_distances = dijkstra(graph, start_node)
shortest_path = shortest_distances['T']
print("S에서 T까지의 최단 경로 정보속도:", shortest_path)
```

```python
import pandas as pd

def floyd_warshall(graph):
    nodes = list(graph.keys())
    n = len(nodes)
    dist = {node: {other: float('inf') for other in nodes} for node in nodes}
    for node in nodes:
        dist[node][node] = 0
    for node in graph:
        for neighbor, weight in graph[node].items():
            dist[node][neighbor] = weight
    for k in nodes:
        for i in nodes:
            for j in nodes:
                dist[i][j] = min(dist[i][j], dist[i][k] + dist[k][j])
    return dist

graph = {
    'S': {'A': 4, 'B': 3, 'C': 5},
    'A': {'S': 4, 'C': 3, 'D': 5},
    'B': {'S': 3, 'C': 1, 'E': 4},
    'C': {'S': 5, 'A': 3, 'B': 1, 'D': 3, 'E': 3},
    'D': {'A': 5, 'C': 3, 'E': 2, 'T': 5},
    'E': {'B': 4, 'C': 3, 'D': 2, 'T': 6},
    'T': {'D': 5, 'E': 6}
}

all_shortest_paths = floyd_warshall(graph)

df = pd.DataFrame(all_shortest_paths)

print("최단 경로 정보속도:")
print(df)
```

```python
from collections import defaultdict, deque

def ford_fulkerson(graph, source, sink):
    residual_graph = defaultdict(dict)
    for u in graph:
        for v, capacity in graph[u]:
            residual_graph[u][v] = capacity
            residual_graph[v][u] = 0 
    
    max_flow = 0 
    max_flow_paths = []
    
    while True:
        parent = {}
        queue = deque([source])
        parent[source] = None
        while queue:
            u = queue.popleft()
            if u == sink:
                break
            for v in residual_graph[u]:
                if v not in parent and residual_graph[u][v] > 0:
                    parent[v] = u
                    queue.append(v)
        
        if sink not in parent:
            break
        
        path_flow = float('inf')
        s = sink
        while s != source:
            path_flow = min(path_flow, residual_graph[parent[s]][s])
            s = parent[s]
        
        max_flow += path_flow
        v = sink
        path = [sink]
        while v != source:
            u = parent[v]
            residual_graph[u][v] -= path_flow
            residual_graph[v][u] += path_flow
            path.append(u)
            v = u
        path.reverse()
        max_flow_paths.append((path, path_flow))
    
    return max_flow, max_flow_paths

graph = {
    'S': [('A', 4), ('B', 3), ('C', 5)],
    'A': [('C', 3), ('D', 5)],
    'B': [('C', 1), ('E', 4)],
    'C': [('D', 3), ('E', 3)],
    'D': [('E', 2), ('T', 5)],
    'E': [('T', 6)],
    'T': []
}

max_flow, max_flow_paths = ford_fulkerson(graph, 'S', 'T')

print("최대 전달 정보량:", max_flow)
print("각 용량별 경로:")
for path, flow in max_flow_paths:
    print(f"{path}: 정보량 {flow}")
```

2.주경로와 최단완성시간을 구하시오.

```python
import networkx as nx

def find_critical_path(graph, completion_times):
    # 역추적을 통해 주경로 찾기
    critical_path = []
    current_vertex = 'S'
    while current_vertex != 'T':
        critical_path.append(current_vertex)
        # 다음으로 이동할 정점은 현재 정점과 연결된 정점 중에서
        # 해당 정점으로 가는 경로가 최단 경로인 정점
        neighbors = list(graph.successors(current_vertex))
        if neighbors:
            current_vertex = max(neighbors, key=lambda x: completion_times[x])
        else:
            break
    critical_path.append('T')
    
    return critical_path

def find_shortest_completion_time(graph):
    # 모든 정점에서부터 최단 경로를 계산
    completion_times = {}
    for vertex in graph.nodes():
        completion_times[vertex] = max(nx.single_source_dijkstra_path_length(graph, vertex).values())
    return completion_times

graph = nx.DiGraph()
graph.add_weighted_edges_from([
    ('S', 'A', 4), ('S', 'B', 3), ('S', 'C', 5),
    ('A', 'C', 3), ('A', 'D', 5),
    ('B', 'C', 1), ('B', 'E', 4),
    ('C', 'D', 3), ('C', 'E', 3),
    ('D', 'E', 2), ('D', 'T', 5),
    ('E', 'T', 6)
])

completion_times = find_shortest_completion_time(graph)
critical_path = find_critical_path(graph, completion_times)

print("주경로:", ' -> '.join(critical_path))
print("최단완성시간:", completion_times['T'])
```
